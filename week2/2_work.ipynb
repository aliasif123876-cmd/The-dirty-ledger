```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("teaching_prices_messy.csv")

# Display the first five rows
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Data Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
cleaned_df = df.copy()
```

```python
# Remove commas from the price column
cleaned_df["price"] = cleaned_df["price"].astype(str).str.replace(",", "", regex=False)

# Convert price column into numeric format
cleaned_df["price"] = pd.to_numeric(
    cleaned_df["price"], errors="coerce"
)

print(cleaned_df["price"].dtype)
```

```python
# Remove unnecessary spaces and standardise capitalization
cleaned_df["item"] = cleaned_df["item"].str.strip()

cleaned_df["item"] = cleaned_df["item"].replace({
    "oil b": "Oil B"
})

print(cleaned_df["item"].value_counts())
```

```python
# Identify duplicate records
print("Duplicates before cleaning:",
      cleaned_df.duplicated().sum())

# Remove exact duplicate records
cleaned_df = cleaned_df.drop_duplicates()

print("Duplicates after cleaning:",
      cleaned_df.duplicated().sum())
```

```python
# Identify missing values
print(cleaned_df.isnull().sum())

# Create a flag for missing prices
cleaned_df["missing_price_flag"] = cleaned_df["price"].isnull()

print(cleaned_df[
    cleaned_df["missing_price_flag"]
])
```

```python
# Identify unusually high prices
cleaned_df["unusual_price_flag"] = (
    cleaned_df["price"] > 1000
)

print(cleaned_df[
    cleaned_df["unusual_price_flag"]
])
```

```python
print("BEFORE CLEANING")
print("Rows:", df.shape[0])
print("Missing Values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())
print("Price Data Type:", df["price"].dtype)

print("\nAFTER CLEANING")
print("Rows:", cleaned_df.shape[0])
print("Missing Values:", cleaned_df["price"].isnull().sum())
print("Duplicates:", cleaned_df.duplicated().sum())
print("Price Data Type:", cleaned_df["price"].dtype)
```
